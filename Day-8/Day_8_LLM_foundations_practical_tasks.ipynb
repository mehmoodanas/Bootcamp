````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Explain the concept of machine learning in simple terms."}],
    max_completion_tokens=100,
    reasoning_effort="none",
)

print(response.choices[0].message.content)
print("Finish reason:", response.choices[0].finish_reason)
print("Output tokens:", response.usage.completion_tokens)
````

````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Explain the concept of machine learning in simple terms."}],
    max_completion_tokens=200,
    reasoning_effort="none",
)

print(response.choices[0].message.content)
print("Finish reason:", response.choices[0].finish_reason)
print("Output tokens:", response.usage.completion_tokens)
````



**50 tokens**

**100 tokens**
The answer stopped at exactly 100 tokens and the finish reason was again "length". It was also cut off in the middle of a sentence, but it had more detail than the 50 token answer.

**200 tokens**
The answer used 193 tokens and the finish reason was "stop". This means the model finished the answer by itself. It was the best answer, with a simple explanation and three steps.

**Summary**
More tokens gave a longer and more detailed answer. But the token limit does not make the answer shorter, it just cuts the text when the limit is reached.

#### 12. Optimize output tokens.

Prompt: `"Why did my movie app recommend a romantic comedy when I mostly watch sci-fi?"`

Rewrite the prompt to request a concise, plausible explanation in 10 output tokens or fewer.

Run the original and revised prompts using the same model and settings. Compare their output token counts and check whether the shorter response still provides a meaningful explanation. Refine your prompt if necessary.

You can use the API response’s usage information to check token counts.

````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Why did my movie app recommend a romantic comedy when I mostly watch sci-fi?"}],
    reasoning_effort="none",
)

print(response.choices[0].message.content)
print("Output tokens:", response.usage.completion_tokens)
````

````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Why did my movie app recommend a romantic comedy when I mostly watch sci-fi? Answer in **maximum 6 words**, no punctuation."}],
    reasoning_effort="none",
)

print(response.choices[0].message.content)
print("Output tokens:", response.usage.completion_tokens)
````

**Conclusion (Task 12):**

**Original prompt**
The model gave a long answer with 259 output tokens. It listed several possible reasons and at the end it offered to explain more. The answer was useful, but much longer than needed.

**Optimized prompt**
I kept the same question and added "Answer in maximum 6 words, no punctuation" in bold. The model answered "Exploration heuristic favored broad appeal", which used only 9 output tokens, so it met the goal of 10 tokens or fewer. That is about 96% fewer output tokens.

**Comparison**
The short answer still gives a meaningful and correct reason, the app tried something different and chose a movie that many people like. But it uses technical words like "heuristic", so a normal user may not understand it as easily as the long answer. The long answer gives more reasons and more detail. Since output tokens cost more than input tokens, asking for a short answer in the prompt is a good way to save money when we only need the main point.

````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Write a Python function called factorial(n) that returns the factorial of n. Raise a ValueError for negative numbers. Return **only the code**, without explanation and without markdown code fences."}],
)

print(response.choices[0].message.content)
````

````python
def factorial(n):
    if n < 0:
        raise ValueError("factorial is not defined for negative numbers")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
print(factorial(0))
print(factorial(1))
print(factorial(5))
print(factorial(10))
try:
    factorial(-3)
except ValueError as e:
    print("Error for -3:", e)
````

#### 23. Extract JSON Data from the input text in two different ways.

Input Text: "Our trip to Tokyo starts on June 12th. We're staying for 5 nights at the Park Hyatt before heading to Kyoto."

Extract the information from the text above using two different methods:

1. Write a prompt asking the AI to return a JSON object with the fields `destination`, `start_date`, and `stay_duration`.

2. Instead of putting rules in the prompt, provide a specific JSON schema that defines the field names and their required data types `destination` (the first city), `start_date` (formatted as YYYY-MM-DD), and `stay_duration` (the number of nights as an integer).

Which method yields a more reliable result for a computer program to process, and why?

````python
import json

text = "Our trip to Tokyo starts on June 12th. We're staying for 5 nights at the Park Hyatt before heading to Kyoto."

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Extract the trip details from this text and return a JSON object with the fields destination, start_date and stay_duration.\n\nText: " + text}],
)

answer = response.choices[0].message.content
print(answer)

try:
    data = json.loads(answer)
    print("Valid JSON. Type of stay_duration:", type(data["stay_duration"]))
except Exception as e:
    print("Could not read as JSON:", e)
````

````python
schema = {
    "type": "object",
    "properties": {
        "destination": {"type": "string", "description": "The first city of the trip"},
        "start_date": {"type": "string", "description": "Start date in YYYY-MM-DD format"},
        "stay_duration": {"type": "integer", "description": "Number of nights"},
    },
    "required": ["destination", "start_date", "stay_duration"],
    "additionalProperties": False,
}

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": "Extract the trip details from this text: " + text}],
    response_format={"type": "json_schema", "json_schema": {"name": "trip_info", "strict": True, "schema": schema}},
)

answer = response.choices[0].message.content
print(answer)

data = json.loads(answer)
print("Type of stay_duration:", type(data["stay_duration"]))
````



**Method 1, prompt only**
The model gave valid JSON, but the values were not good for a program. The date was "June 12th" and the stay was "5 nights", which is text, not a number. A program needs extra code to fix this.

**Method 2, JSON schema**
The model gave the exact fields with the correct types. The date was "2026-06-12" and the stay was the number 5. My code showed the type is int.

**Which method is more reliable**
The JSON schema is more reliable. With a prompt only, the model can choose its own types, add extra text or break the JSON. With a schema, the output must follow the exact fields and types, so a program can always read it.

  3. Create evaluation criteria for measuring relevance, completeness, tone, and factual reliability.
- **Interactive Assignment:** Run three versions of a research or writing prompt, score the outputs with a rubric, and document the revisions that produced the strongest result.
