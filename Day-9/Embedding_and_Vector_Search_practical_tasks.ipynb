Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python
# complete the code
sentences_list = [
    "I am attending my bootcamp nowadays and I learn every day new topics with some hands on experience from our mentors.",
    "The first day of bootcamp was offline on the campus, and it was fun meeting the colleagues working with us in our bootcamp.",
    "On the first day we learned data cleaning: how to clean and sort data so the dataset looks good and is easy to read.",
    "Artificial intelligence is a very interesting and vast topic to study. You can go as deep as you want but never reach its end.",
    "Our mentors are very good and helpful. They teach us, guide us through our tasks, and answer any question we ask them.",
    "I want to work very hard to show everyone that I am capable of doing everything and that I can achieve my goals.",
    "Maybe some days will be bad and I will fall, but I will stand up again, boost my energy and start from the beginning.",
    "Today we studied RAG, vector search and vector databases. I will try to do the tasks and learn something new today.",
    "Sometimes I feel very happy because I am enjoying doing all of this, and the topics we study seem very interesting to me.",
    "Today I also have to go to work, so I will finish my tasks quickly to reach work on time and support my colleagues.",
]    #your sentences
embeddings = model.encode(sentences_list)
print(embeddings.shape)

```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
# place your code here
def cos_similarity_calculation(text, sentences_list):

    text_embedding = model.encode(text)

    list_embeddings = model.encode(sentences_list)

    similarities = util.cos_sim(text_embedding, list_embeddings)[0]

    for sentence, score in zip(sentences_list, similarities):
        print(f"{score.item():.3f}  {sentence}")

    return similarities

result = cos_similarity_calculation("What did we learn on the first day of the bootcamp?", sentences_list)
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
# place your code here
import numpy as np

def cosine_similarity_batch(vector, batch):
    dot_products = np.dot(batch, vector)

    norms = np.linalg.norm(batch, axis=1) * np.linalg.norm(vector)
    return dot_products / norms

# Demonstration: one query vector vs the batch of my sentence vectors
query_vector = model.encode("Which topics about vector databases did we study today?")
scores = cosine_similarity_batch(query_vector, embeddings)

for sentence, score in zip(sentences_list, scores):
    print(f"{score:.3f}  {sentence}")

# The highest score is the most similar sentence
best = np.argmax(scores)
print("\nMost similar sentence:", sentences_list[best])
```

* Cosine similarity measures how close two sentences are in meaning. Two sentences about the bootcamp scored 0.61, while unrelated sentences scored close to 0 (for example -0.006). Cosine distance is 1 minus this similarity, so a smaller distance means more similar meaning.
* My function 'cos_similarity_calculation' found the sentences most related to a new question.
